```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Entorno configurado correctamente")
```

```python
datos = pd.DataFrame({
    "mes": ["Enero", "Febrero", "Marzo", "Abril"],
    "ventas": [120, 150, 135, 180]
})

datos
```

```python
sns.barplot(data=datos, x="mes", y="ventas")
plt.title("Ventas por mes")
plt.show()
```
